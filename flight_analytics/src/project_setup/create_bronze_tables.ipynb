```python
%sql
DROP TABLE IF EXISTS dev_flight_analytics.bronze.flight_events; 
CREATE EXTERNAL TABLE IF NOT EXISTS dev_flight_analytics.bronze.flight_events (
    event_type STRING,
    flight_id STRING,
    flight_number STRING,
    old_status STRING,
    new_status STRING,
    delay_minutes INT,
    gate STRING,
    event_time TIMESTAMP,
    partition_key STRING,
    sequence_number LONG,
    offset LONG,
    message_id STRING,
    enqueued_time TIMESTAMP,
    created_time TIMESTAMP
)
USING DELTA
CLUSTER BY (event_type, flight_id)
LOCATION "s3://ragesh-databricks-s3-inputs/flight_analytics/dev/bronze/flight_events";
```
